```python

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

flights_data = pd.read_csv('flights.csv')
weather_data_pd = pd.read_csv('weather.csv')
weather_data_np = weather_data_pd.to_numpy()
```

```python
flights_data.head(10)
```

```python
#Question 1 How many flights were there from JFK to SLC?
q_1 = flights_data[(flights_data['origin'] == 'JFK') & (flights_data['dest'] == 'SLC')].shape[0]
print(q_1)
```

```python
#Question 2 How many airlines fly to SLC? Should be int
q_2 = flights_data[(flights_data['dest'] == 'SLC')].shape[0]
print(q_2)
```

```python
#Question 3 What is the average arrival delay for flights to RDU?
q_3 = flights_data[flights_data['dest'] == 'RDU']['arr_delay'].mean()
print(q_3)
```

```python
#Question 4 What proportion of flights to SEA come from the two NYC airports (LGA and JFK)?

total_SEA_flights = flights_data[(flights_data['dest'] == 'SEA')].shape[0]


total_SEA_flights_from_JFK_or_LGA = flights_data[
    ((flights_data['origin'] == 'JFK') | (flights_data['origin'] == 'LGA')) & 
    (flights_data['dest'] == 'SEA')
].shape[0]

q_4 = total_SEA_flights_from_JFK_or_LGA / total_SEA_flights

print(q_4)
```

```python
#Question 5 Which date has the largest average depature delay? Pd slice with date and float

processed_flights = flights_data.copy()

processed_flights['date_formatted'] = (
    processed_flights['year'].astype(str) + '/' + 
    processed_flights['month'].astype(str) + '/' + 
    processed_flights['day'].astype(str)
)

q_5 = processed_flights.groupby('date_formatted')['dep_delay'].mean().nlargest(1).to_frame(name='average_dep_delay')

display(q_5)

```

```python
#Question 6 Which date has the largest average arrival delay? pd slice with date and float
q_6 = processed_flights.groupby('date_formatted')['arr_delay'].mean().nlargest(1).to_frame(name='average_arr_delay')

display(q_6)

```

```python
#Question 7 Which flight departing LGA or JFK in 2013 flew the fastest? pd slice with tailnumber and speed speed = distance/airtime
filtered_LGA_JFK_departures = processed_flights[
    (processed_flights['origin'] == 'JFK') | (processed_flights['origin'] == 'LGA')
]

filtered_LGA_JFK_departures['speed_mph'] = (filtered_LGA_JFK_departures['distance'] / filtered_LGA_JFK_departures['air_time'])

q_7 = filtered_LGA_JFK_departures.nsmallest(1, 'speed_mph')[['tailnum', 'speed_mph']]

print(q_7)

```

```python
#Question 9 How many observations were made in Feburary?

q_9 = q_8[(q_8['month'] == 2.0)].shape[0]

print(q_9)
```

```python
#Question 10 What was the mean for humidity in February? Float

q_10 = q_8[(q_8['month'] == 2.0)]['humid'].mean()

print(q_10)
```

```python
#Question 11 What was the std for humidity in February? Float

q_11 = q_8[(q_8['month'] == 2.0)]['humid'].std()

print(q_11)
```
